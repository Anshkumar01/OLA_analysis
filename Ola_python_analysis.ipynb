```python

import pandas as pd
df = pd.read_csv(r'Bookings_ola.csv')
```

```python
df.head()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Driver_Ratings'] = df.groupby('Vehicle_Type')['Driver_Ratings'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['Customer_Rating'] = df.groupby('Vehicle_Type')['Customer_Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['V_TAT'] = df.groupby('Vehicle_Type')['V_TAT'].transform(lambda x: x.fillna(x.median()))

df['C_TAT'] = df.groupby('Vehicle_Type')['C_TAT'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns
```

```python
df.head()
```

```python
df = df.drop('vehicle images', axis=1)
```

```python
df.columns
```

```python
df.drop('unnamed: 20', axis=1, inplace=True)
```

```python
df.columns
```

```python
df.loc[df['booking_status'] != 'Success', 'payment_method'] = 'No Payment'
```
